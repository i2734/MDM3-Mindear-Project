```python
#!/usr/bin/env python3
"""
SoundTrackr Smartphone Calibration Data Organiser & Error Analysis Pipeline
---------------------------------------------------------------------------
Complete, self-contained script compatible with:
  - Updated SoundTrackr Collector schema (15 Nov 2026: 49 columns, true IEC 61672
    A/C/Z weighting, `clipped` flags, and bundled multi-trial CSV exports)
  - Legacy 15-column Collector exports
  - SoundTracker_Calibration_Kit folder structure & SoundTracker_Recording_Names.xlsx
  - Charlie Experiment Variances (Section 3: energy averaging, pooled bias & variance,
    repeatability breakdown, normality check, and seeded simulation error helper)
  - Jupyter Notebook, Spyder, IPython, and standard terminal execution
"""

from __future__ import annotations

import argparse
import io
import json
import re
from pathlib import Path
from typing import Dict, Optional, Tuple

import matplotlib.pyplot as plt
import numpy as np
import pandas as pd
from scipy import stats


# ==============================================================================
# CONFIGURATION CONSTANTS
# ==============================================================================
# Number of seconds to retain per trial (central window if recording is longer).
# Set to 10 (or 20 for full 20 s app trials, or None to use all valid unclipped rows).
TARGET_WINDOW_SECONDS: Optional[int] = 10

# Default random seed for reproducible simulated measurement error generation
DEFAULT_RANDOM_SEED = 42

# Nominal target levels (dBA) used as fallback when no reference meter log is entered
CONDITION_TARGET_DB: Dict[str, float] = {
    "E1_65dB": 65.0,
    "E1_80dB": 80.0,
    "E1_95dB": 95.0,
    "E2_90deg": 80.0,
    "E2_180deg": 80.0,
    "E3_Pocket": 80.0,
}


# ==============================================================================
# CORE ACOUSTIC & STATISTICAL FUNCTIONS
# ==============================================================================
def energy_mean_db(
    levels_db: pd.Series | np.ndarray,
    durations_s: Optional[pd.Series | np.ndarray] = None,
) -> float:
    """
    Calculate equivalent continuous sound level (Leq) via energy averaging.
    Never averages decibels arithmetically (Charlie Experiment Variances, Eq. 1).
    """
    levels = np.asarray(levels_db, dtype=float)
    valid_mask = np.isfinite(levels)
    levels = levels[valid_mask]

    if levels.size == 0:
        return np.nan

    linear_energy = np.power(10.0, levels / 10.0)

    if durations_s is not None:
        durations = np.asarray(durations_s, dtype=float)[valid_mask]
        if np.all(np.isfinite(durations)) and np.sum(durations) > 0:
            mean_energy = np.sum(durations * linear_energy) / np.sum(durations)
            return float(10.0 * np.log10(mean_energy))

    return float(10.0 * np.log10(np.mean(linear_energy)))


def apply_measurement_error(
    l_true: np.ndarray | pd.Series | float,
    mu: float,
    sigma: float,
    seed: int = DEFAULT_RANDOM_SEED,
) -> np.ndarray:
    """
    Apply experimentally informed measurement uncertainty to simulated readings
    (Charlie Experiment Variances, Section 3.4, Eqs. 6-7):
        L_measured,i = L_true,i + mu + eps_i,   eps_i ~ N(0, sigma^2)
    """
    rng = np.random.default_rng(seed)
    true_arr = np.asarray(l_true, dtype=float)
    epsilon = rng.normal(loc=0.0, scale=sigma, size=true_arr.shape)
    return true_arr + mu + epsilon


# ==============================================================================
# ROBUST CSV READER & MULTI-TRIAL BUNDLE INDEXER
# ==============================================================================
def smart_read_collector_csv(csv_path: Path) -> pd.DataFrame:
    """
    Read a SoundTrackr Collector CSV robustly:
      - Handles comma (,) and semicolon (;) delimiters
      - Skips any metadata/comment lines above the real table header
      - Normalises column names case-insensitively to canonical Collector names
    """
    with open(csv_path, "r", encoding="utf-8-sig", errors="replace") as f:
        raw_text = f.read()

    if not raw_text or not raw_text.strip():
        return pd.DataFrame()

    lines = raw_text.splitlines()
    header_idx = 0
    sep = ","
    for i, line in enumerate(lines[:20]):
        low = line.lower()
        if any(tok in low for tok in ("laeq", "recorded_at", "recording_id", "event_id")):
            header_idx = i
            sep = ";" if line.count(";") > line.count(",") else ","
            break
    else:
        first_line = next((l for l in lines if l.strip()), "")
        sep = ";" if first_line.count(";") > first_line.count(",") else ","

    df = pd.read_csv(io.StringIO("\n".join(lines[header_idx:])), sep=sep)
    df.columns = [str(c).replace("\ufeff", "").strip() for c in df.columns]

    canon_map = {}
    for c in df.columns:
        cl = c.lower().strip().replace(" ", "_").replace("(", "").replace(")", "")
        if cl in ("laeq_db", "laeq", "laeq_dba", "laeq_1s", "dba", "level_db", "sound_level_db"):
            canon_map[c] = "laeq_db"
        elif cl in ("lceq_db", "lceq", "lceq_dbc"):
            canon_map[c] = "lceq_db"
        elif cl in ("lzeq_db", "lzeq", "lzeq_dbz"):
            canon_map[c] = "lzeq_db"
        elif cl in ("recorded_at", "timestamp", "time", "utc_time", "datetime"):
            canon_map[c] = "recorded_at"
        elif cl in ("recording_id", "recordingid"):
            canon_map[c] = "recording_id"
        elif cl in ("event_id", "eventid"):
            canon_map[c] = "event_id"
        elif cl in ("clipped", "is_clipped"):
            canon_map[c] = "clipped"
        elif cl in ("clipped_fraction", "clip_fraction"):
            canon_map[c] = "clipped_fraction"

    return df.rename(columns=canon_map)


def _canonical_condition(text: str) -> Optional[str]:
    s_low = str(text or "").lower()
    if "65db" in s_low:
        return "E1_65dB"
    if "95db" in s_low:
        return "E1_95dB"
    if "90deg" in s_low:
        return "E2_90deg"
    if "180deg" in s_low:
        return "E2_180deg"
    if "pocket" in s_low or "mfront" in s_low:
        return "E3_Pocket"
    if "80db" in s_low:
        return "E1_80dB"
    return None


def _infer_cond_and_phone(
    text: str, fallback_cond: str = "", fallback_phone: str = ""
) -> Tuple[Optional[str], Optional[str]]:
    s = (text or "").strip()
    m_phone = re.search(
        r"(?:^|[^A-Za-z0-9])(IP\d{2}|SA\d{2}|[A-Z]{2}\d{2})(?:$|[^A-Za-z0-9])",
        s,
        re.IGNORECASE,
    )
    phone_id = (
        m_phone.group(1).upper()
        if m_phone
        else (fallback_phone.upper() if re.match(r"^(IP|SA|[A-Z]{2})\d{2}$", fallback_phone, re.I) else None)
    )

    cond = _canonical_condition(s) or _canonical_condition(fallback_cond)
    return cond, phone_id


def _extract_trial_num(text: str) -> Optional[int]:
    if not text:
        return None
    s = str(text).strip()
    m = re.search(
        r"(?:^|[^A-Za-z0-9])(?:T|trial[_\s-]?)0?(\d+)(?:$|[^A-Za-z0-9])",
        s,
        re.IGNORECASE,
    )
    if m:
        return int(m.group(1))
    m2 = re.search(r"(?:_|-)0?([1-5])$", s)
    if m2:
        return int(m2.group(1))
    return None


def build_bundled_trial_index(
    raw_root: Path,
) -> Dict[Tuple[str, str, int], Tuple[Path, str, str, str, str]]:
    """
    Scan all CSVs inside 01_Raw_Data/ and map each (Condition_Folder, Phone_ID, Trial)
    to (csv_path, filter_col, filter_val, start_utc, end_utc) without overwriting
    already-matched trial slots.
    """
    index: Dict[Tuple[str, str, int], Tuple[Path, str, str, str, str]] = {}
    if not raw_root.exists():
        return index

    csv_files = [p for p in sorted(raw_root.rglob("*.csv")) if not p.name.startswith("._")]
    print(f"Scanning {len(csv_files)} raw CSV file(s) in {raw_root}...")

    def _claim_slot(
        c: Optional[str], p: Optional[str], preferred_t: Optional[int], entry: Tuple[Path, str, str, str, str]
    ) -> bool:
        if not (c and p):
            return False
        if preferred_t is not None and (c, p, preferred_t) not in index:
            index[(c, p, preferred_t)] = entry
            return True
        for cand in range(1, 11):
            if (c, p, cand) not in index:
                index[(c, p, cand)] = entry
                return True
        return False

    for csv_path in csv_files:
        path_parts = [csv_path.stem, csv_path.parent.name, csv_path.parent.parent.name]
        fallback_phone = next((x for x in path_parts if re.match(r"^(IP|SA|[A-Z]{2})\d{2}$", x, re.I)), "")
        fallback_cond = next((x for x in path_parts if x in CONDITION_TARGET_DB), "")

        try:
            df = smart_read_collector_csv(csv_path)
        except Exception as exc:
            print(f"[WARNING] Could not read {csv_path}: {exc}")
            continue

        if df.empty:
            continue

        added_this_file = 0

        # Strategy 1: Match by recording_id, recording_started_at, or event_id
        for id_col in ("recording_id", "recording_started_at", "event_id"):
            if id_col not in df.columns or not df[id_col].notna().any():
                continue

            id_counts = df[id_col].value_counts()
            if "recorded_at" in df.columns:
                tmp = df[[id_col, "recorded_at"]].dropna(subset=[id_col]).copy()
                tmp["_dt"] = pd.to_datetime(tmp["recorded_at"], utc=True, errors="coerce")
                ordered_ids = tmp.groupby(id_col, sort=False)["_dt"].min().sort_values().index.tolist()
            else:
                ordered_ids = df[id_col].dropna().unique().tolist()

            # Ignore < 5 s aborted false-starts when full trials are present
            full_ids = [rid for rid in ordered_ids if id_counts.get(rid, 0) >= 5]
            candidate_ids = full_ids if len(full_ids) >= 1 else ordered_ids

            # 1a. Explicit trial number inside the ID (e.g. T01..T05)
            explicit_hits = 0
            for raw_id in candidate_ids:
                c, p = _infer_cond_and_phone(f"{csv_path.stem} {raw_id}", fallback_cond, fallback_phone)
                t = _extract_trial_num(str(raw_id))
                if c and p and t is not None:
                    if _claim_slot(c, p, t, (csv_path, id_col, str(raw_id), "", "")):
                        added_this_file += 1
                    explicit_hits += 1
            if explicit_hits > 0:
                break

            # 1b. Chronological recording_ids / recording_started_at -> assign next open trial slot 1..5
            if id_col in ("recording_id", "recording_started_at"):
                for raw_id in candidate_ids:
                    c, p = _infer_cond_and_phone(f"{csv_path.stem} {raw_id}", fallback_cond, fallback_phone)
                    if _claim_slot(c, p, None, (csv_path, id_col, str(raw_id), "", "")):
                        added_this_file += 1
                if added_this_file > 0:
                    break

        if added_this_file > 0:
            continue

        # Strategy 2: Split multi-trial file by time gaps (> 3 seconds between recordings)
        c, p = _infer_cond_and_phone(csv_path.stem, fallback_cond, fallback_phone)
        if c and p and "recorded_at" in df.columns:
            dt = pd.to_datetime(df["recorded_at"], utc=True, errors="coerce").dropna().sort_values()
            if len(dt) > 0:
                burst_ids = (dt.diff().dt.total_seconds().fillna(0) > 3.0).cumsum() + 1
                if burst_ids.nunique() > 1:
                    for _, grp in dt.groupby(burst_ids):
                        if len(grp) < 5 and burst_ids.nunique() > 5:
                            continue
                        if _claim_slot(c, p, None, (csv_path, "", "", grp.min().isoformat(), grp.max().isoformat())):
                            added_this_file += 1
                    if added_this_file > 0:
                        continue

        # Strategy 3: Single-trial file fallback
        if c and p:
            t = _extract_trial_num(csv_path.stem)
            if _claim_slot(c, p, t, (csv_path, "", "", "", "")):
                added_this_file += 1

        if added_this_file == 0:
            print(f"[WARNING] 0 trials extracted from: {csv_path} (cols={df.columns.tolist()[:8]})")

    # Report any (Condition, Phone) pairs that still have fewer than 5 trials indexed
    for cond in CONDITION_TARGET_DB:
        for phone in sorted({k[1] for k in index}):
            count = sum(1 for (c, p, t) in index if c == cond and p == phone and 1 <= t <= 5)
            if count < 5:
                print(f"[INFO] {cond} / {phone}: matched {count}/5 trials")

    return index


# ==============================================================================
# DATA LOADING, CLEANING & ORGANISATION
# ==============================================================================
def load_experiment_manifest(kit_root: Path) -> Tuple[pd.DataFrame, pd.DataFrame]:
    xlsx_path = kit_root / "SoundTracker_Recording_Names.xlsx"
    if not xlsx_path.exists():
        raise FileNotFoundError(f"Could not find manifest workbook at: {xlsx_path}")

    xl = pd.ExcelFile(xlsx_path)
    rec_sheet = "Recording Names" if "Recording Names" in xl.sheet_names else xl.sheet_names[0]
    recordings_df = pd.read_excel(xlsx_path, sheet_name=rec_sheet)

    if "Phone ID Key" in xl.sheet_names:
        phone_key_df = pd.read_excel(xlsx_path, sheet_name="Phone ID Key")
    else:
        phone_key_df = pd.DataFrame(columns=["Phone ID", "Phone type", "Actual model (fill in)"])

    recordings_df.columns = [str(c).strip() for c in recordings_df.columns]
    phone_key_df.columns = [str(c).strip() for c in phone_key_df.columns]

    # Locate real header row dynamically if Excel has a banner row at the top
    if not any("filename" in c.lower() for c in recordings_df.columns):
        for idx in range(min(10, len(recordings_df))):
            row_vals = [str(v).strip() for v in recordings_df.iloc[idx].tolist()]
            if any("filename" in v.lower() for v in row_vals):
                recordings_df.columns = row_vals
                recordings_df = recordings_df.iloc[idx + 1 :].reset_index(drop=True)
                break

    fname_col = next(
        (c for c in recordings_df.columns if "filename" in c.lower()),
        recordings_df.columns[0],
    )

    recordings_df = recordings_df.dropna(how="all").copy()
    fname_s = recordings_df[fname_col].astype(str).str.strip()
    valid_mask = (
        recordings_df[fname_col].notna()
        & (fname_s != "")
        & (fname_s.str.lower() != "nan")
    )
    recordings_df = recordings_df[valid_mask].copy().reset_index(drop=True)
    recordings_df["Filename (copy to rename)"] = recordings_df[fname_col].astype(str).str.strip()

    if recordings_df.empty:
        raise ValueError(
            f"No valid recording rows found in '{rec_sheet}' of {xlsx_path}. "
            f"Detected columns: {recordings_df.columns.tolist()}"
        )

    col_map = {c.lower().strip(): c for c in recordings_df.columns}

    recordings_df["Condition_Folder"] = (
        recordings_df["Filename (copy to rename)"]
        .apply(lambda x: _canonical_condition(x) or "Unknown_Condition")
    )

    phone_col = next((v for k, v in col_map.items() if "phone" in k and "id" in k), None)
    if phone_col is not None:
        recordings_df["Phone ID"] = recordings_df[phone_col].astype(str).str.strip()
    else:
        recordings_df["Phone ID"] = (
            recordings_df["Filename (copy to rename)"]
            .str.extract(r"(?:^|_)([A-Za-z]{2}\d{2})(?:_|\.|$)", expand=False)
            .fillna("UNKNOWN")
        )
    
    # Swap template SA03 -> IP04 to match actual hardware recorded in 01_Raw_Data
    recordings_df["Phone ID"] = recordings_df["Phone ID"].replace({"SA03": "IP04"})
    recordings_df["Filename (copy to rename)"] = recordings_df["Filename (copy to rename)"].str.replace(
        "SA03", "IP04", regex=False
    )
    if "Phone ID" in phone_key_df.columns:
        phone_key_df["Phone ID"] = phone_key_df["Phone ID"].replace({"SA03": "IP04"})
        if "Phone type" in phone_key_df.columns:
            phone_key_df.loc[phone_key_df["Phone ID"] == "IP04", "Phone type"] = "Apple"

    trial_col = next(
        (v for k, v in col_map.items() if k in ("trial", "trial_id", "trial id", "repeat", "run")),
        None,
    )
    if trial_col is not None:
        extracted = recordings_df[trial_col].astype(str).str.extract(r"(\d+)", expand=False)
        trial_series = pd.to_numeric(extracted, errors="coerce")
    else:
        trial_series = pd.Series(np.nan, index=recordings_df.index)

    fname_trial = recordings_df["Filename (copy to rename)"].str.extract(r"(?:_|-)T0?(\d+)", expand=False)
    trial_series = trial_series.fillna(pd.to_numeric(fname_trial, errors="coerce"))
    if trial_series.isna().any():
        auto_nums = recordings_df.groupby(["Condition_Folder", "Phone ID"]).cumcount() + 1
        trial_series = trial_series.fillna(auto_nums)
    recordings_df["Trial"] = trial_series.astype(int)

    exp_col = next((v for k, v in col_map.items() if "experiment" in k), None)
    recordings_df["Experiment"] = recordings_df[exp_col].fillna("Calibration") if exp_col else "Calibration"

    lvl_col = next((v for k, v in col_map.items() if "target" in k or "level" in k), None)
    if lvl_col:
        raw_lvl = recordings_df[lvl_col].astype(str).str.extract(r"(\d+(?:\.\d+)?)", expand=False)
        recordings_df["Target level (dBA)"] = pd.to_numeric(raw_lvl, errors="coerce")
    else:
        recordings_df["Target level (dBA)"] = pd.to_numeric(
            recordings_df["Filename (copy to rename)"].str.extract(r"(\d+)\s*dB", expand=False),
            errors="coerce",
        )
    recordings_df["Target level (dBA)"] = recordings_df["Target level (dBA)"].fillna(
        recordings_df["Condition_Folder"].map(CONDITION_TARGET_DB)
    )

    ori_col = next((v for k, v in col_map.items() if "orient" in k or "angle" in k), None)
    if ori_col:
        recordings_df["Orientation"] = recordings_df[ori_col].fillna("0°").astype(str)
    else:
        recordings_df["Orientation"] = (
            recordings_df["Filename (copy to rename)"]
            .str.extract(r"(90deg|180deg|0deg)", expand=False)
            .fillna("0°")
        )

    pkt_col = next((v for k, v in col_map.items() if "pocket" in k), None)
    if pkt_col:
        recordings_df["Pocket"] = recordings_df[pkt_col].fillna("No").astype(str)
    else:
        recordings_df["Pocket"] = np.where(
            recordings_df["Filename (copy to rename)"].str.lower().str.contains("pocket"), "Yes", "No"
        )

    recordings_df["Expected_Rel_Path"] = (
        "01_Raw_Data/"
        + recordings_df["Condition_Folder"].astype(str)
        + "/"
        + recordings_df["Phone ID"].astype(str)
        + "/"
        + recordings_df["Filename (copy to rename)"].astype(str)
    )
    return recordings_df, phone_key_df


def load_or_create_reference_log(
    kit_root: Path, recordings_df: pd.DataFrame
) -> pd.DataFrame:
    ref_dir = kit_root / "04_Reference_Meter_Exports"
    ref_dir.mkdir(parents=True, exist_ok=True)
    ref_log_path = ref_dir / "Reference_Log.csv"

    if ref_log_path.exists():
        ref_df = pd.read_csv(ref_log_path)
        for str_col in ["Source_Export_File", "Window_Start_UTC", "Window_End_UTC", "Notes"]:
            if str_col in ref_df.columns:
                ref_df[str_col] = ref_df[str_col].fillna("").astype(str)
    else:
        ref_df = pd.DataFrame(
            {
                "Filename": recordings_df["Filename (copy to rename)"],
                "Condition_Folder": recordings_df["Condition_Folder"],
                "Experiment": recordings_df["Experiment"],
                "Target_Level_dBA": recordings_df["Target level (dBA)"],
                "Orientation": recordings_df["Orientation"],
                "Pocket": recordings_df["Pocket"],
                "Phone_ID": recordings_df["Phone ID"],
                "Trial": recordings_df["Trial"],
                "Ref_LAeq_dB": np.nan,
                "Ref_Before_dB": np.nan,
                "Ref_After_dB": np.nan,
                "Source_Export_File": "",
                "Window_Start_UTC": "",
                "Window_End_UTC": "",
                "Notes": "",
            }
        )
        ref_df.to_csv(ref_log_path, index=False)
        print(f"[INFO] Created template reference log at: {ref_log_path}")

    for num_col in ("Ref_LAeq_dB", "Ref_Before_dB", "Ref_After_dB"):
        if num_col in ref_df.columns:
            ref_df[num_col] = pd.to_numeric(ref_df[num_col], errors="coerce")
        else:
            ref_df[num_col] = np.nan

    cond_series = (
        ref_df["Condition_Folder"]
        if "Condition_Folder" in ref_df.columns
        else ref_df["Filename"].astype(str).str.extract(r"^([^_]+_[^_]+)", expand=False)
    )

    target_raw = (
        ref_df["Target_Level_dBA"].astype(str).str.extract(r"(\d+(?:\.\d+)?)", expand=False)
        if "Target_Level_dBA" in ref_df.columns
        else pd.Series(np.nan, index=ref_df.index)
    )
    target_num = pd.to_numeric(target_raw, errors="coerce").fillna(
        cond_series.map(CONDITION_TARGET_DB)
    )

    before_after_mean = ref_df[["Ref_Before_dB", "Ref_After_dB"]].mean(axis=1)
    measured_ref = ref_df["Ref_LAeq_dB"].fillna(before_after_mean)

    ref_df["Reference_Source"] = np.where(
        measured_ref.notna(),
        "Reference_SLM",
        "Target_Level_Fallback (±2 dB tolerance)",
    )
    ref_df["Resolved_Ref_LAeq_dB"] = measured_ref.fillna(target_num)
    return ref_df


def clean_and_extract_trial(
    raw_csv_path: Path,
    filter_col: str = "",
    filter_val: str = "",
    window_start_utc: str = "",
    window_end_utc: str = "",
    target_window_seconds: Optional[int] = TARGET_WINDOW_SECONDS,
) -> Tuple[pd.DataFrame, dict]:
    df = smart_read_collector_csv(raw_csv_path)
    if "laeq_db" not in df.columns:
        raise ValueError(
            f"Missing required 'laeq_db' column in {raw_csv_path}. "
            f"Columns actually found: {df.columns.tolist()}"
        )

    # Isolate trial rows when reading from a bundled multi-trial CSV
    if filter_col and filter_col in df.columns and filter_val:
        df = df[df[filter_col].astype(str) == str(filter_val)].copy()

    # Sort chronologically by ISO-8601 UTC window-end timestamp
    if "recorded_at" in df.columns:
        df["recorded_at_dt"] = pd.to_datetime(df["recorded_at"], utc=True, errors="coerce")
        df = df.sort_values("recorded_at_dt").reset_index(drop=True)

    def _is_valid_ts(val: object) -> bool:
        return (
            val is not None
            and pd.notna(val)
            and str(val).strip().lower() not in ("", "nan", "none", "nat")
        )

    has_start = _is_valid_ts(window_start_utc)
    has_end = _is_valid_ts(window_end_utc)
    if (has_start or has_end) and "recorded_at_dt" in df.columns:
        if has_start:
            df = df[df["recorded_at_dt"] >= pd.to_datetime(str(window_start_utc).strip(), utc=True)]
        if has_end:
            df = df[df["recorded_at_dt"] <= pd.to_datetime(str(window_end_utc).strip(), utc=True)]
        df = df.reset_index(drop=True)

    n_raw_rows = len(df)

    # Convert acoustic columns to float (handles both '.' and ',' decimal separators)
    for lvl_col in ("laeq_db", "lceq_db", "lzeq_db"):
        if lvl_col in df.columns:
            cleaned_num = df[lvl_col].astype(str).str.strip().str.replace(",", ".", regex=False)
            df[lvl_col] = pd.to_numeric(cleaned_num, errors="coerce")

    # Updated app (15 Nov 2026): drop seconds flagged `clipped` (no 30-120 dB clamp)
    clipped_mask = pd.Series(False, index=df.index)
    if "clipped" in df.columns:
        clipped_mask = clipped_mask | df["clipped"].astype(str).str.lower().isin(["true", "1", "yes"])
    if "clipped_fraction" in df.columns:
        cf_str = df["clipped_fraction"].astype(str).str.strip().str.replace(",", ".", regex=False)
        cf = pd.to_numeric(cf_str, errors="coerce").fillna(0.0)
        clipped_mask = clipped_mask | (cf > 0.0)

    n_clipped_dropped = int(clipped_mask.sum())
    clean_df = df[~clipped_mask & df["laeq_db"].notna()].copy().reset_index(drop=True)

    # Trim to uniform analysis window if longer
    if (
        target_window_seconds is not None
        and not (has_start and has_end)
        and len(clean_df) > target_window_seconds
    ):
        trim_start = (len(clean_df) - target_window_seconds) // 2
        clean_df = clean_df.iloc[trim_start : trim_start + target_window_seconds].reset_index(drop=True)

    def _first_non_null(col: str) -> str:
        if col in clean_df.columns and clean_df[col].dropna().shape[0] > 0:
            return str(clean_df[col].dropna().iloc[0])
        if col in df.columns and df[col].dropna().shape[0] > 0:
            return str(df[col].dropna().iloc[0])
        return ""

    diagnostics = {
        "n_raw_rows": n_raw_rows,
        "n_clipped_dropped": n_clipped_dropped,
        "n_valid_rows": len(clean_df),
        "contributor_bucket": _first_non_null("contributor_bucket"),
        "phone_model": _first_non_null("phone_model"),
        "device_class": _first_non_null("device_class"),
        "browser": _first_non_null("browser"),
        "os": _first_non_null("os"),
        "holding": _first_non_null("holding"),
        "trim_db": _first_non_null("trim_db") or _first_non_null("calibration_offset_db"),
        "reference_offset_version": _first_non_null("reference_offset_version"),
        "recording_id": _first_non_null("recording_id"),
        "mic_agc": _first_non_null("mic_auto_gain_control"),
        "mic_ns": _first_non_null("mic_noise_suppression"),
        "mic_ec": _first_non_null("mic_echo_cancellation"),
    }
    return clean_df.drop(columns=["recorded_at_dt"], errors="ignore"), diagnostics


# ==============================================================================
# MAIN PROCESSING & STATISTICAL ANALYSIS WORKFLOW
# ==============================================================================
def run_pipeline(kit_root: Path, random_seed: int = DEFAULT_RANDOM_SEED) -> dict:
    raw_dir = kit_root / "01_Raw_Data"
    processed_dir = kit_root / "02_Processed_Data"
    analysis_dir = kit_root / "03_Analysis"
    processed_dir.mkdir(parents=True, exist_ok=True)
    analysis_dir.mkdir(parents=True, exist_ok=True)

    # 1. Load naming manifest, reference log, and index all raw CSVs
    recordings_df, phone_key_df = load_experiment_manifest(kit_root)
    ref_df = load_or_create_reference_log(kit_root, recordings_df)
    bundled_index = build_bundled_trial_index(raw_dir)

    recordings_df["_join_key"] = (
        recordings_df["Filename (copy to rename)"].astype(str).str.strip().str.lower().str.removesuffix(".csv")
    )
    ref_df["_join_key"] = (
        ref_df["Filename"].astype(str).str.strip().str.lower().str.removesuffix(".csv")
    )

    merged_schedule = pd.merge(
        recordings_df,
        ref_df[
            [
                "_join_key",
                "Resolved_Ref_LAeq_dB",
                "Reference_Source",
                "Source_Export_File",
                "Window_Start_UTC",
                "Window_End_UTC",
                "Notes",
            ]
        ],
        on="_join_key",
        how="left",
    )

    fallback_ref = pd.to_numeric(merged_schedule["Target level (dBA)"], errors="coerce").fillna(
        merged_schedule["Condition_Folder"].map(CONDITION_TARGET_DB)
    )
    merged_schedule["Resolved_Ref_LAeq_dB"] = merged_schedule["Resolved_Ref_LAeq_dB"].fillna(fallback_ref)
    merged_schedule["Reference_Source"] = merged_schedule["Reference_Source"].fillna(
        "Target_Level_Fallback (±2 dB tolerance)"
    )

    # 2. Process all scheduled trials
    trial_records = []
    discovered_devices = {}

    for _, row in merged_schedule.iterrows():
        fname = row["Filename (copy to rename)"]
        if not fname.lower().endswith(".csv"):
            fname = f"{fname}.csv"
        cond_folder = row["Condition_Folder"]
        phone_id = row["Phone ID"]
        trial_num = int(row["Trial"])

        exact_path = kit_root / row["Expected_Rel_Path"]
        filter_col, filter_val, auto_start, auto_end = "", "", "", ""
        raw_path: Optional[Path] = None

        if exact_path.exists():
            raw_path = exact_path
        elif (cond_folder, phone_id, trial_num) in bundled_index:
            raw_path, filter_col, filter_val, auto_start, auto_end = bundled_index[
                (cond_folder, phone_id, trial_num)
            ]
        elif isinstance(row.get("Source_Export_File"), str) and row["Source_Export_File"].strip():
            candidate = raw_dir / row["Source_Export_File"].strip()
            if candidate.exists():
                raw_path = candidate

        start_utc = str(row.get("Window_Start_UTC", "") or "") or auto_start
        end_utc = str(row.get("Window_End_UTC", "") or "") or auto_end

        record = {
            "Filename": fname,
            "Condition_Folder": cond_folder,
            "Experiment": row["Experiment"],
            "Target_Level_dBA": row["Target level (dBA)"],
            "Orientation": row["Orientation"],
            "Pocket": row["Pocket"],
            "Phone_ID": phone_id,
            "Trial": trial_num,
            "Raw_File_Found": raw_path is not None,
            "Source_CSV": raw_path.name if raw_path else "",
            "N_Raw_Rows": 0,
            "N_Clipped_Dropped": 0,
            "N_Valid_Seconds": 0,
            "Phone_LAeq_dB": np.nan,
            "Phone_LCeq_dB": np.nan,
            "Phone_LZeq_dB": np.nan,
            "Reference_LAeq_dB": row["Resolved_Ref_LAeq_dB"],
            "Reference_Source": row["Reference_Source"],
            "Measurement_Error_dB": np.nan,
            "Contributor_Bucket": "",
            "Phone_Model": "",
            "Device_Class": "",
            "Browser": "",
            "Holding": "",
            "Trim_dB": "",
            "Offset_Version": "",
            "QC_Status": "Missing Raw File",
        }

        if raw_path is not None:
            clean_df, diag = clean_and_extract_trial(
                raw_csv_path=raw_path,
                filter_col=filter_col,
                filter_val=filter_val,
                window_start_utc=start_utc,
                window_end_utc=end_utc,
                target_window_seconds=TARGET_WINDOW_SECONDS,
            )

            clean_df.to_csv(processed_dir / fname, index=False)

            durations = (
                clean_df["integration_seconds"]
                if "integration_seconds" in clean_df.columns
                else None
            )
            l_phone = energy_mean_db(clean_df["laeq_db"], durations_s=durations)
            l_c = (
                energy_mean_db(clean_df["lceq_db"], durations_s=durations)
                if "lceq_db" in clean_df.columns
                else np.nan
            )
            l_z = (
                energy_mean_db(clean_df["lzeq_db"], durations_s=durations)
                if "lzeq_db" in clean_df.columns
                else np.nan
            )

            l_ref = row["Resolved_Ref_LAeq_dB"]
            err = (l_phone - l_ref) if (np.isfinite(l_phone) and np.isfinite(l_ref)) else np.nan

            record.update(
                {
                    "N_Raw_Rows": diag["n_raw_rows"],
                    "N_Clipped_Dropped": diag["n_clipped_dropped"],
                    "N_Valid_Seconds": diag["n_valid_rows"],
                    "Phone_LAeq_dB": round(l_phone, 4) if np.isfinite(l_phone) else np.nan,
                    "Phone_LCeq_dB": round(l_c, 4) if np.isfinite(l_c) else np.nan,
                    "Phone_LZeq_dB": round(l_z, 4) if np.isfinite(l_z) else np.nan,
                    "Measurement_Error_dB": round(err, 4) if np.isfinite(err) else np.nan,
                    "Contributor_Bucket": diag["contributor_bucket"],
                    "Phone_Model": diag["phone_model"],
                    "Device_Class": diag["device_class"],
                    "Browser": diag["browser"],
                    "Holding": diag["holding"],
                    "Trim_dB": diag["trim_db"],
                    "Offset_Version": diag["reference_offset_version"],
                    "QC_Status": (
                        "Valid"
                        if np.isfinite(err)
                        else ("Missing Reference LAeq" if np.isfinite(l_phone) else "All Rows Clipped/Invalid")
                    ),
                }
            )

            if phone_id not in discovered_devices and diag["phone_model"]:
                discovered_devices[phone_id] = diag

        trial_records.append(record)

    trials_df = pd.DataFrame(trial_records)
    trials_df.to_csv(processed_dir / "All_Trials_Processed.csv", index=False)

    # Save updated Device Register
    device_reg = phone_key_df.copy()
    for out_col, diag_key in [
        ("Detected_Phone_Model", "phone_model"),
        ("Contributor_Bucket", "contributor_bucket"),
        ("Device_Class", "device_class"),
        ("Browser", "browser"),
        ("OS", "os"),
        ("Mic_AGC", "mic_agc"),
        ("Mic_Noise_Suppression", "mic_ns"),
        ("Mic_Echo_Cancellation", "mic_ec"),
    ]:
        device_reg[out_col] = device_reg["Phone ID"].map(
            lambda pid, k=diag_key: discovered_devices.get(pid, {}).get(k, "")
        )
    device_reg.to_csv(processed_dir / "Device_Register_Updated.csv", index=False)

    valid_df = trials_df[trials_df["Measurement_Error_dB"].notna()].copy()
    m_valid = len(valid_df)

    print(f"Total scheduled trials: {len(trials_df)}")
    print(f"Raw trials matched:     {int(trials_df['Raw_File_Found'].sum())}")
    print(f"Valid paired trials:    {m_valid}")
    print("QC breakdown:          ", trials_df["QC_Status"].value_counts().to_dict())

    if m_valid < 2:
        print(
            "\n[ACTION REQUIRED] Fewer than 2 valid paired trials found.\n"
            "Check 02_Processed_Data/All_Trials_Processed.csv for per-row QC_Status details."
        )
        return {"valid_count": m_valid}

    # 3. Overall Bias (mu) and Pooled Sample Variance (sigma^2) — Eqs. 3, 4, 5
    errors = valid_df["Measurement_Error_dB"].to_numpy(dtype=float)
    mu = float(np.mean(errors))
    sigma_sq = float(np.var(errors, ddof=1))
    sigma = float(np.sqrt(sigma_sq))

    # 4. Breakdowns by Condition, Phone, and Phone x Condition Repeatability
    by_condition = (
        valid_df.groupby(
            ["Condition_Folder", "Experiment", "Target_Level_dBA", "Orientation", "Pocket"],
            as_index=False,
        )
        .agg(
            N_Trials=("Measurement_Error_dB", "count"),
            Mean_Error_dB=("Measurement_Error_dB", "mean"),
            Variance_dB2=("Measurement_Error_dB", lambda s: np.var(s, ddof=1) if len(s) > 1 else np.nan),
            SD_dB=("Measurement_Error_dB", lambda s: np.std(s, ddof=1) if len(s) > 1 else np.nan),
            Total_Clipped_Seconds=("N_Clipped_Dropped", "sum"),
        )
    )
    by_condition.to_csv(analysis_dir / "Summary_By_Condition.csv", index=False)

    by_phone = (
        valid_df.groupby(["Phone_ID"], as_index=False)
        .agg(
            N_Trials=("Measurement_Error_dB", "count"),
            Mean_Error_dB=("Measurement_Error_dB", "mean"),
            Variance_dB2=("Measurement_Error_dB", lambda s: np.var(s, ddof=1) if len(s) > 1 else np.nan),
            SD_dB=("Measurement_Error_dB", lambda s: np.std(s, ddof=1) if len(s) > 1 else np.nan),
            Total_Clipped_Seconds=("N_Clipped_Dropped", "sum"),
        )
    )
    by_phone.to_csv(analysis_dir / "Summary_By_Phone.csv", index=False)

    repeatability_df = (
        valid_df.groupby(["Condition_Folder", "Phone_ID"], as_index=False)
        .agg(
            N_Repeats=("Measurement_Error_dB", "count"),
            Cell_Mean_Error_dB=("Measurement_Error_dB", "mean"),
            Repeatability_Var_dB2=(
                "Measurement_Error_dB",
                lambda s: np.var(s, ddof=1) if len(s) > 1 else np.nan,
            ),
            Repeatability_SD_dB=(
                "Measurement_Error_dB",
                lambda s: np.std(s, ddof=1) if len(s) > 1 else np.nan,
            ),
        )
    )
    repeatability_df.to_csv(analysis_dir / "Summary_Repeatability_By_Cell.csv", index=False)

    valid_cells = repeatability_df[repeatability_df["N_Repeats"] > 1]
    if not valid_cells.empty:
        dof_rep = (valid_cells["N_Repeats"] - 1).sum()
        pooled_rep_var = float(
            ((valid_cells["N_Repeats"] - 1) * valid_cells["Repeatability_Var_dB2"]).sum() / dof_rep
        )
        pooled_rep_sd = float(np.sqrt(pooled_rep_var))
    else:
        pooled_rep_var, pooled_rep_sd = np.nan, np.nan

    shapiro_w, shapiro_p = (
        stats.shapiro(errors) if 3 <= m_valid <= 5000 else (np.nan, np.nan)
    )

    overall_summary = {
        "valid_recordings_M": m_valid,
        "overall_bias_mu_db": round(mu, 4),
        "pooled_variance_sigma2_db2": round(sigma_sq, 4),
        "pooled_sd_sigma_db": round(sigma, 4),
        "pure_repeatability_variance_db2": round(pooled_rep_var, 4) if np.isfinite(pooled_rep_var) else None,
        "pure_repeatability_sd_db": round(pooled_rep_sd, 4) if np.isfinite(pooled_rep_sd) else None,
        "shapiro_wilk_W": round(float(shapiro_w), 4) if np.isfinite(shapiro_w) else None,
        "shapiro_wilk_p": round(float(shapiro_p), 4) if np.isfinite(shapiro_p) else None,
        "recommended_random_seed": random_seed,
    }
    with open(analysis_dir / "Overall_Error_Parameters.json", "w", encoding="utf-8") as f:
        json.dump(overall_summary, f, indent=2)
    pd.DataFrame([overall_summary]).to_csv(analysis_dir / "Overall_Error_Parameters.csv", index=False)

    # 5. Histogram of Pooled Measurement Errors vs. Fitted Normal Distribution
    fig, ax = plt.subplots(figsize=(8.5, 5.2), dpi=150)
    n_bins = min(20, max(8, int(np.sqrt(m_valid))))
    ax.hist(
        errors,
        bins=n_bins,
        density=True,
        alpha=0.65,
        color="#2b6cb0",
        edgecolor="white",
        label=f"Pooled Experimental Errors (M = {m_valid})",
    )

    x_min = min(errors.min(), mu - 3.5 * sigma)
    x_max = max(errors.max(), mu + 3.5 * sigma)
    x_grid = np.linspace(x_min, x_max, 300)
    ax.plot(
        x_grid,
        stats.norm.pdf(x_grid, loc=mu, scale=sigma),
        color="#c53030",
        linewidth=2.2,
        label=rf"Fitted Normal $\mathcal{{N}}(\mu={mu:.2f},\ \sigma^2={sigma_sq:.2f})$",
    )
    ax.axvline(mu, color="#c53030", linestyle="--", linewidth=1.3, label=rf"Mean Bias $\mu = {mu:.2f}$ dB")
    ax.axvline(0.0, color="#2d3748", linestyle=":", linewidth=1.2, label="Zero Error (0 dB)")

    stats_box = (
        f"M = {m_valid} recordings\n"
        f"Bias (μ) = {mu:+.3f} dB\n"
        f"Variance (σ²) = {sigma_sq:.3f} dB²\n"
        f"Std Dev (σ) = {sigma:.3f} dB\n"
        f"Shapiro-Wilk p = {shapiro_p:.3f}"
    )
    ax.text(
        0.03,
        0.95,
        stats_box,
        transform=ax.transAxes,
        fontsize=9.5,
        verticalalignment="top",
        bbox=dict(boxstyle="round,pad=0.4", facecolor="#f7fafc", edgecolor="#cbd5e0"),
    )

    ax.set_title("Pooled Smartphone Measurement Error Distribution vs. Fitted Normal", fontsize=12, fontweight="bold")
    ax.set_xlabel(r"Measurement Error $e_i = L_{\mathrm{phone}, i} - L_{\mathrm{reference}, i}$ (dB)", fontsize=10.5)
    ax.set_ylabel("Probability Density", fontsize=10.5)
    ax.grid(axis="y", alpha=0.3)
    ax.legend(loc="upper right", fontsize=9)
    fig.tight_layout()
    fig.savefig(analysis_dir / "Pooled_Error_Histogram_vs_Normal.png")
    plt.show()
    plt.close(fig)

    print("\n=== OVERALL CALIBRATION RESULTS ===")
    for k, v in overall_summary.items():
        print(f"  {k}: {v}")
    print(f"\nOutputs saved to:\n  - {processed_dir}\n  - {analysis_dir}")

    return overall_summary


if __name__ == "__main__":
    parser = argparse.ArgumentParser(
        description="Process SoundTrackr calibration recordings (15 Nov 2026 schema)."
    )
    parser.add_argument(
        "--kit-dir",
        type=Path,
        default=Path("SoundTracker_Calibration_Kit"),
        help="Path to the SoundTracker_Calibration_Kit root directory.",
    )
    parser.add_argument(
        "--seed",
        type=int,
        default=DEFAULT_RANDOM_SEED,
        help="Random seed for reproducible simulation error draws.",
    )
    args, _ = parser.parse_known_args()
    run_pipeline(kit_root=args.kit_dir, random_seed=args.seed)
```
